```python
import pandas as pd
```

```python
sales=pd.read_csv("Sales.csv", encoding="latin-1")
cust=pd.read_csv("Customers.csv", encoding="latin-1")
products=pd.read_csv("Products.csv",encoding="latin-1")
stores=pd.read_csv("Stores.csv", encoding="latin-1")
```

```python
#Fix date type

sales["Order Date"]=pd.to_datetime(sales["Order Date"],format= "%m/%d/%Y")
for col in ["Unit Cost USD", "Unit Price USD"]:
    products[col]=products[col].astype(str).str.replace(r"[$,\s]","",regex=True).astype(float)
```

```python
# 3. Merge everything onto Sales (one join per lookup table)

df=(sales
    .merge(cust, on="CustomerKey", how="left", validate="many_to_one")
    .merge(products, on="ProductKey", how="left", validate="many_to_one")
    .merge(stores.rename(columns={"Country":"Store Country"}),
           on="StoreKey",how="left", validate="many_to_one"))
```

```python
# 4. Create the new columns
df["Revenue USD"]=df["Quantity"] * df["Unit Price USD"]
df["Profit"]=df["Quantity"] *(df["Unit Price USD"] - df["Unit Cost USD"])
df["Sales Channel"]=df["StoreKey"].apply(lambda k: "Online" if k== 0 else "In store")
```
